## Paso 3 · Prueba funcional integrada

Esta celda ejecuta una operación real con cada paquete sobre datos sintéticos.  
Confirma que no solo se importan, sino que funcionan correctamente juntos.

````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
import statsmodels.formula.api as smf
from statsmodels.stats.power import TTestIndPower
from sklearn.decomposition import PCA
import pingouin as pg

rng = np.random.default_rng(seed=42)
errores = []

# 1. pandas — crear DataFrame
try:
    n = 80
    df = pd.DataFrame({
        "grupo":       np.repeat(["A", "B"], n // 2),
        "experiencia": rng.integers(1, 15, n),
        "satisfaccion": rng.normal(7, 1.5, n).clip(1, 10).round(1),
        "productividad": rng.normal(75, 12, n).round(1),
    })
    print("✅ pandas   — DataFrame de 80 filas × 4 columnas creado.")
except Exception as e:
    errores.append("pandas"); print(f"❌ pandas: {e}")

# 2. scipy — Shapiro-Wilk
try:
    stat, p = stats.shapiro(df["productividad"])
    print(f"✅ scipy    — Shapiro-Wilk: W={stat:.4f}, p={p:.4f}")
except Exception as e:
    errores.append("scipy"); print(f"❌ scipy: {e}")

# 3. statsmodels — OLS + potencia
try:
    modelo = smf.ols("productividad ~ experiencia + satisfaccion", data=df).fit()
    print(f"✅ statsmodels — OLS ajustado: R²={modelo.rsquared:.4f}")
    analisis = TTestIndPower()
    n_req = analisis.solve_power(effect_size=0.5, alpha=0.05, power=0.80)
    print(f"   └─ Potencia estadística: n requerido (d=0.5, α=.05, 1-β=.80) = {n_req:.0f}")
except Exception as e:
    errores.append("statsmodels"); print(f"❌ statsmodels: {e}")

# 4. scikit-learn — PCA
try:
    X = df[["experiencia", "satisfaccion", "productividad"]].values
    pca = PCA(n_components=2)
    pca.fit(X)
    var_exp = pca.explained_variance_ratio_.sum()
    print(f"✅ scikit-learn — PCA (2 componentes): varianza explicada={var_exp:.1%}")
except Exception as e:
    errores.append("scikit-learn"); print(f"❌ scikit-learn: {e}")

# 5. pingouin — d de Cohen
try:
    d = pg.compute_effsize(
        df.loc[df.grupo=="A", "productividad"],
        df.loc[df.grupo=="B", "productividad"],
        eftype="cohen"
    )
    print(f"✅ pingouin  — d de Cohen entre grupos A y B: d={d:.4f}")
except Exception as e:
    errores.append("pingouin"); print(f"❌ pingouin: {e}")

# 6. matplotlib + seaborn — figura
try:
    fig, axes = plt.subplots(1, 2, figsize=(10, 4))
    sns.histplot(df["productividad"], kde=True, ax=axes[0], color="steelblue")
    axes[0].set_title("Distribución: productividad")
    axes[0].set_xlabel("Productividad")
    axes[0].set_ylabel("Frecuencia")
    sns.boxplot(x="grupo", y="productividad", data=df, ax=axes[1],
                palette=["#4C72B0", "#DD8452"])
    axes[1].set_title("Productividad por grupo")
    axes[1].set_xlabel("Grupo")
    axes[1].set_ylabel("Productividad")
    fig.suptitle("Prueba funcional — entorno_verificacion.ipynb", fontsize=12)
    plt.tight_layout()
    plt.savefig("verificacion_figura.png", dpi=100, bbox_inches="tight")
    plt.show()
    print("✅ matplotlib + seaborn — figura generada y guardada como verificacion_figura.png")
except Exception as e:
    errores.append("matplotlib/seaborn"); print(f"❌ matplotlib/seaborn: {e}")

print(f"\nErrores detectados: {len(errores)}")
if errores:
    print("   Paquetes con error:", ", ".join(errores))
````
